```python
import re
import json
import dateparser
from datetime import datetime, timedelta
```

```python
mock_calendar = {
    "Monday": ["09:00", "10:00", "14:00", "15:00"],
    "Tuesday": ["11:00", "13:00", "16:00"],
    "Wednesday": ["09:00", "14:00", "17:00"],
    "Thursday": ["10:00", "15:00"],
    "Friday": ["09:00", "11:00", "14:00"]
}
```

```python
meeting_request = input("Enter your meeting request: ")
```

```python
import re

duration_match = re.search(r'(\d+)\s*(hour|hours|minute|minutes)', meeting_request.lower())

if duration_match:
    value = int(duration_match.group(1))
    unit = duration_match.group(2)

    if "hour" in unit:
        duration_minutes = value * 60
    else:
        duration_minutes = value

else:
    duration_minutes = 30

print(duration_minutes)
print(type(duration_minutes))
```

```python
if "morning" in meeting_request.lower():
    time_preference = "morning"
elif "afternoon" in meeting_request.lower():
    time_preference = "afternoon"
elif "evening" in meeting_request.lower():
    time_preference = "evening"
else:
    time_preference = "No preference"

print(time_preference)
```

```python
parsed_date = dateparser.parse(meeting_request)

print(parsed_date)
```

```python
if "next week" in meeting_request.lower():
    day_preference = "next week"
elif "tomorrow" in meeting_request.lower():
    day_preference = "tomorrow"
elif "today" in meeting_request.lower():
    day_preference = "today"
else:
    day_preference = "Not specified"

print(day_preference)
```

```python
candidate_slots = []

for day, times in mock_calendar.items():
    for time in times:
        hour = int(time[:2])

        if 13 <= hour <= 17:
            candidate_slots.append(f"{day} {time}")

print(candidate_slots)
```

```python
print("Suggested Meeting Slots:\n")

for slot in candidate_slots[:3]:
    print(slot)
```

```python
from datetime import datetime, timedelta

time = "14:00"

start_time = datetime.strptime(time, "%H:%M")

end_time = start_time + timedelta(minutes=duration_minutes)

print("Start Time:", start_time.strftime("%H:%M"))
print("End Time:", end_time.strftime("%H:%M"))
```

```python
print("\nSuggested Meeting Slots:")

for i, slot in enumerate(candidate_slots[:3], start=1):

    day, time = slot.split()

    start_time = datetime.strptime(time, "%H:%M")
    end_time = start_time + timedelta(minutes=duration_minutes)

    print(
        f"{i}. {day} "
        f"{start_time.strftime('%H:%M')} - "
        f"{end_time.strftime('%H:%M')}"
    )
```
